```python
import pandas as pd

routes = pd.read_csv("../dataset/raw/kolkata/gtfs/routes.txt")

print(routes.columns)
print(routes.head())
print(routes["route_type"].value_counts())
```

```python
metro_routes = routes[routes["route_type"] == 1]

print("Total Metro Routes:", len(metro_routes))

display(metro_routes)
```

```python
train_routes = routes[routes["route_type"] == 2]

print("Total Railway Routes:", len(train_routes))

display(train_routes)
```

```python
bus_routes = routes[routes["route_type"] == 3]

print("Total Bus Routes:", len(bus_routes))

display(bus_routes.head(10))
```

```python
metro_routes.to_csv(
    "../dataset/processed/metro_routes.csv",
    index=False
)

train_routes.to_csv(
    "../dataset/processed/train_routes.csv",
    index=False
)

bus_routes.to_csv(
    "../dataset/processed/bus_routes.csv",
    index=False
)
```
