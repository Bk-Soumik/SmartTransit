```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "../dataset/simulated/smarttransit_simulated_hourly_passengers_2025.csv"
)

print("Original Dataset Shape:", df.shape)
df.isnull().sum()
print("Duplicate rows:", df.duplicated().sum())
```

```python
df_clean = df.drop_duplicates().copy()

print("Shape after removing duplicates:", df_clean.shape)
```

```python
df_clean["datetime"] = pd.to_datetime(df_clean["datetime"])
df_clean["date"] = pd.to_datetime(df_clean["date"])

print(df_clean.dtypes)
```

```python
print("Negative passenger counts:",
      (df_clean["passenger_count"] < 0).sum())

print("Negative capacities:",
      (df_clean["assumed_hourly_capacity"] < 0).sum())
```

```python
print(df_clean["occupancy_percentage"].describe())
```

```python
df_clean.to_csv(
    "../dataset/processed/smarttransit_cleaned_2025.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
